# Human Activity Recognition — EDA
Exploring the UCI HAR dataset: class balance, feature separability and raw signals.

> Run `python src/download_data.py` first.

```python
import sys; sys.path.append('../src')
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from data_loader import load_features, load_signals, activity_labels
labels = activity_labels()
X, y, subj = load_features('train')
X.shape
```

```python
names = [labels[i+1] for i in y]
sns.countplot(y=names, order=list(labels.values()))
plt.title('Class distribution (train)'); plt.show()
```

```python
df = pd.DataFrame({'activity': names, 'tBodyAccMag-mean()': X['tBodyAccMag-mean()']})
sns.boxplot(data=df, x='tBodyAccMag-mean()', y='activity')
plt.title('Static vs dynamic activities separate on acc magnitude'); plt.show()
```

```python
from sklearn.manifold import TSNE
emb = TSNE(n_components=2, random_state=42, init='pca').fit_transform(X.sample(2000, random_state=42))
idx = X.sample(2000, random_state=42).index
sns.scatterplot(x=emb[:,0], y=emb[:,1], hue=np.array(names)[idx], s=8)
plt.title('t-SNE of 561 features'); plt.legend(bbox_to_anchor=(1,1)); plt.show()
```

```python
S, ys = load_signals('train')
fig, axes = plt.subplots(2, 3, figsize=(14,6), sharey=True)
for ax, c in zip(axes.flat, range(6)):
    i = np.where(ys == c)[0][0]
    ax.plot(S[i,:,0:3]); ax.set_title(labels[c+1])
fig.suptitle('Body acceleration (x,y,z) — one window per activity'); plt.tight_layout(); plt.show()
```
